# PHREEQC validation

Compare the compositional PHREEQC workflow against a separate reference implementation over the injection benchmark. Windows MATLAB and a registered IPhreeqcCOM server are required.

```matlab
startupH2sim
% Set this path to your separate reference checkout.
referenceRoot = 'C:/Projects/UGFACT';
result = runPhreeqcValidation( ...
    'ugfactRoot', referenceRoot, 'injectionDays', 50);

plotCompositionalPhreeqcValidation(result, 'outputDirectory', ...
    fullfile(pwd, 'build', 'validation-figures'));
```

## Injection benchmark



The domain contains 50 cells over 50 m. Both simulations use a common Soreide–Whitson phase model and the moderate kinetic rates of the reference setup. The chemistry splitting cadence differs, so comparison errors must be reported rather than assuming identical curves. Full setups and accepted outputs are saved.

Source attribution: the reference implementation is [UGFACT](https://github.com/ahmadrezashojaee/UGFACT), by Ahmadreza Shojaee. The bundled database source revision and credits are documented under `h2-biochem/database/README.md`.

The dissolved-H2 plateau agrees closely after matching the EOS. The pH front and reaction histories differ. The reference consumption curve is an end-rate time integral; H2sim accumulates accepted reaction amounts. The 50-day diagnostics are approximately 27.9 and 42.8 mol, respectively. This benchmark does not establish timestep convergence or matching kinetic histories. Small sulfur-balance residuals exceed the configured audit tolerance.
