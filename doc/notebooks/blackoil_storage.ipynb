# Black-oil hydrogen–brine displacement

Run a small horizontal injection model with the `h2-store` module. This tutorial is serial, works without PHREEQC, and downloads no data. The oil phase represents brine; dissolution and vaporization are disabled.

```matlab
startupH2sim
result = exampleBlackOilH2Storage1D('gridCells', 40, 'numSteps', 12);

% Inspect the final accepted state.
state = result.states{end};
gasSaturation = state.s(:, 2);
pressureMPa = state.pressure / 1e6;
```

## Read the profiles



The hydrogen gas density uses the local Brill–Beggs correlation. Constant brine density and phase viscosities are tutorial assumptions. The setup returns its model, initial state, schedule, well history, and accepted states.

Change `gridCells` or `numSteps` to explore resolution and injection duration. This model has an open pressure boundary and no withdrawal cycle; it is not a field case.
