# 2D hydrogen storage in an aquifer

Use the existing structural-trap aquifer model for hydrogen build-up and five storage cycles. Gravity, capillarity, and tabulated dissolution are active. The bundled deck uses zero added salt; diffusion and PHREEQC are disabled.

```matlab
startupH2sim
result = exampleBlackOilAquifer2D('numCycles', 5);

% Redraw saved accepted states.
plotBlackOilAquifer2D(result, 'outputDirectory', ...
    fullfile(pwd, 'build', 'aquifer-figures'));
state = result.states{end};
gasSaturation = state.s(:, 2);
solutionRatio = state.rs;
```

## Plume and well response





The 50 m × 50 m grid has a 2 m target spacing. The 500-day schedule includes 120 days of build-up, a 30-day shut-in, and five 70-day cycles. Each cycle contains 30 days of injection, 10 days idle, and 30 days withdrawal. Plume panels share a color scale. `result.packedProblem` identifies the setup and accepted outputs on disk.
